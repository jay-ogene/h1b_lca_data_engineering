```python
import numpy as np  
def annualize_wage(value, unit):
    if pd.isnull(value) or pd.isnull(unit):
        return np.nan
    unit = str(unit).lower()
    if unit == 'year':
        return value
    elif unit == 'hour':
        return value * 2080
    elif unit == 'week':
        return value * 52
    elif unit == 'month':
        return value * 12
    elif 'bi' in unit:  # Handles "Bi-Weekly"
        return value * 26
    else:
        return np.nan  # Invalid unit


# Annualize offered wages
df_clean['WAGE_RATE_FROM_ANNUAL'] = df_clean.apply(
    lambda x: annualize_wage(x['WAGE_RATE_OF_PAY_FROM'], x['WAGE_UNIT_OF_PAY']), axis=1
)
df_clean['WAGE_RATE_TO_ANNUAL'] = df_clean.apply(
    lambda x: annualize_wage(x['WAGE_RATE_OF_PAY_TO'], x['WAGE_UNIT_OF_PAY']), axis=1
)

# Annualize prevailing wages
df_clean['PREVAILING_WAGE_ANNUAL'] = df_clean.apply(
    lambda x: annualize_wage(x['PREVAILING_WAGE'], x['PW_UNIT_OF_PAY']), axis=1
)
```

```python
# Calculate mean wage
df_clean['WAGE_RATE_MEAN'] = (df_clean['WAGE_RATE_FROM_ANNUAL'] + df_clean['WAGE_RATE_TO_ANNUAL']) / 2

# Flag non-compliant wages (offered < prevailing)
df_clean['BELOW_PREVAILING_WAGE'] = df_clean['WAGE_RATE_MEAN'] < df_clean['PREVAILING_WAGE_ANNUAL']
```

```python
import pandas as pd

# Define the columns we'll keep in the employers table
employer_columns = [
    'EMPLOYER_NAME', 
    'EMPLOYER_STATE', 
    'EMPLOYER_REGION',
    'EMPLOYER_CITY',
    'NAICS_CODE', 
    'INDUSTRY_NAME'
    # Add more if relevant
]

# Create df_employers by selecting just employer columns, then drop duplicates
df_employers = df_final[employer_columns].drop_duplicates().reset_index(drop=True)

# Create a surrogate employer_id
df_employers['employer_id'] = df_employers.index + 1  # simple integer sequence

# Reorder columns to have employer_id first
cols = ['employer_id'] + employer_columns
df_employers = df_employers[cols]

df_employers.head()
```

```python
# 1. Merge df_final with df_employers to get employer_id
merge_cols = employer_columns  # same columns you used to define uniqueness
tmp = df_final.merge(df_employers, on=merge_cols, how='left')

# 2. Pick columns for job_postings
job_posting_columns = [
    'CASE_NUMBER',
    'JOB_TITLE',
    'employer_id',      # from the merge
    'SOC_CODE',
    'WAGE_RATE_MEAN',
    'WORKSITE_STATE',
    'WORKSITE_REGION',
    'VISA_CLASS',
    'BEGIN_DATE',
    'END_DATE',
    # plus any other columns relevant at the posting level
]


df_job_postings = tmp[job_posting_columns].drop_duplicates(subset=['CASE_NUMBER'])
# Usually CASE_NUMBER is unique anyway, but we use drop_duplicates just in case.

df_job_postings.head()
```

```python
from sqlalchemy import create_engine, text

engine = create_engine("postgresql://myuser:mypassword@localhost:5432/h1b_db")

with engine.connect() as conn:
    # Drop if exist for clean re-run
    conn.execute(text("DROP TABLE IF EXISTS job_postings;"))
    conn.execute(text("DROP TABLE IF EXISTS employers;"))

    # Create employers table
    conn.execute(text("""
        CREATE TABLE employers (
            employer_id    INT     PRIMARY KEY,
            employer_name  TEXT    NOT NULL,
            employer_state TEXT,
            employer_region TEXT,
            naics_code     TEXT,
            industry_name  TEXT
        );
    """))
    
    # Create job_postings table
    conn.execute(text("""
        CREATE TABLE job_postings (
            case_number     TEXT    PRIMARY KEY,
            employer_id     INT     REFERENCES employers(employer_id),
            soc_code        TEXT,
            wage_rate_mean  FLOAT,
            worksite_state  TEXT,
            worksite_region TEXT,
            begin_date      DATE,
            end_date        DATE
        );
    """))
```

```python
df_employers.to_sql(
    'employers',
    con=engine,
    if_exists='append',   # we already created the table
    index=False           # we don't want a separate index column
)
```

```python
df_job_postings.to_sql(
    'job_postings',
    con=engine,
    if_exists='append',
    index=False
)
```

```python
from sqlalchemy import text

query = text('''
    SELECT e."INDUSTRY_NAME", COUNT(*)
    FROM job_postings jp
    JOIN employers e ON jp.employer_id = e.employer_id
    GROUP BY e."INDUSTRY_NAME"
    ORDER BY COUNT(*) DESC
    LIMIT 10;
''')

with engine.connect() as conn:
    rows = conn.execute(query).fetchall()

for r in rows:
    print(r)
```
